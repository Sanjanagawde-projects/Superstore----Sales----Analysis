```python
import numpy as np
import pandas as pd
df = pd.read_csv("/content/Sample - Superstore.csv", encoding='latin1')
df.head()
df.shape
df.columns
df.info()
df.describe()
df["Order Date"] = pd.to_datetime(df["Order Date"], format='mixed')
df["Ship Date"] = pd.to_datetime(df["Ship Date"], format='mixed')
df["Sales"].mean()
```

```python
sales_by_cateory = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)
print(sales_by_cateory)
```

```python
Sales_by_region = df.groupby("Region")["Sales"].sum().sort_values(ascending=False)
print(Sales_by_region)
```

```python
Sales_by_Segment=df.groupby("Segment")["Sales"].sum().sort_values(ascending=False)
print(Sales_by_Segment)
```

```python
Top_Customers=(
    df.groupby("Customer Name")["Sales"].sum().sort_values(ascending=False).head(10)
)
print(Top_Customers)
```

```python
Top_Products=(
    df.groupby("Product Name")["Sales"].sum().sort_values(ascending=False).head(5)
)
Top_Products
```

```python
loss_products=(df.groupby("Sub-Category")["Profit"].sum().sort_values())
loss_products
```

```python
avg_discount=df.groupby("Region")["Discount"].mean().sort_values(ascending=False)
avg_discount
```
